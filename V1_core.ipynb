```python
# =========================================================
# PULSEPREDICT
# Train/Test Split + Preprocessing + ML Algorithms
# =========================================================

import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

import joblib


# =========================================================
# 1. LOAD CLEAN DATA
# =========================================================



print("Dataset shape:", df.shape)


# =========================================================
# 2. SEPARATE FEATURES AND TARGET
# =========================================================

X = df.drop("target", axis=1)

y = df["target"]

print("\nX shape:", X.shape)
print("y shape:", y.shape)


# =========================================================
# 3. DEFINE NUMERICAL FEATURES
# =========================================================

numerical_features = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]


# =========================================================
# 4. DEFINE CATEGORICAL FEATURES
# =========================================================

categorical_features = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]


# =========================================================
# 5. TRAIN / TEST SPLIT
# =========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# =========================================================
# 6. CREATE PREPROCESSING PIPELINE
# =========================================================

preprocessor = ColumnTransformer(
    transformers=[
        
        # Numerical columns
        (
            "numerical",
            StandardScaler(),
            numerical_features
        ),
        
        # Categorical columns
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)


# =========================================================
# 7. LOGISTIC REGRESSION PIPELINE
# =========================================================

logistic_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        (
            "model",
            LogisticRegression(
                max_iter=1000,
                random_state=42
            )
        )
    ]
)


# =========================================================
# 8. TRAIN LOGISTIC REGRESSION
# =========================================================

print("\nTraining Logistic Regression...")

logistic_pipeline.fit(
    X_train,
    y_train
)

print("Logistic Regression training complete.")


# =========================================================
# 9. LOGISTIC REGRESSION PREDICTIONS
# =========================================================

lr_predictions = logistic_pipeline.predict(X_test)

lr_probabilities = logistic_pipeline.predict_proba(
    X_test
)[:, 1]


# =========================================================
# 10. LOGISTIC REGRESSION EVALUATION
# =========================================================

print("\n")
print("=" * 60)
print("LOGISTIC REGRESSION RESULTS")
print("=" * 60)

print(
    "Accuracy:",
    accuracy_score(y_test, lr_predictions)
)

print(
    "Precision:",
    precision_score(y_test, lr_predictions)
)

print(
    "Recall:",
    recall_score(y_test, lr_predictions)
)

print(
    "F1 Score:",
    f1_score(y_test, lr_predictions)
)

print(
    "ROC-AUC:",
    roc_auc_score(y_test, lr_probabilities)
)

print(
    "PR-AUC:",
    average_precision_score(
        y_test,
        lr_probabilities
    )
)

print("\nConfusion Matrix:")

print(
    confusion_matrix(
        y_test,
        lr_predictions
    )
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        lr_predictions
    )
)


# =========================================================
# 11. RANDOM FOREST PIPELINE
# =========================================================

random_forest_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        (
            "model",
            RandomForestClassifier(
                n_estimators=200,
                random_state=42,
                class_weight="balanced"
            )
        )
    ]
)


# =========================================================
# 12. TRAIN RANDOM FOREST
# =========================================================

print("\nTraining Random Forest...")

random_forest_pipeline.fit(
    X_train,
    y_train
)

print("Random Forest training complete.")


# =========================================================
# 13. RANDOM FOREST PREDICTIONS
# =========================================================

rf_predictions = random_forest_pipeline.predict(
    X_test
)

rf_probabilities = random_forest_pipeline.predict_proba(
    X_test
)[:, 1]


# =========================================================
# 14. RANDOM FOREST EVALUATION
# =========================================================

print("\n")
print("=" * 60)
print("RANDOM FOREST RESULTS")
print("=" * 60)

print(
    "Accuracy:",
    accuracy_score(y_test, rf_predictions)
)

print(
    "Precision:",
    precision_score(y_test, rf_predictions)
)

print(
    "Recall:",
    recall_score(y_test, rf_predictions)
)

print(
    "F1 Score:",
    f1_score(y_test, rf_predictions)
)

print(
    "ROC-AUC:",
    roc_auc_score(y_test, rf_probabilities)
)

print(
    "PR-AUC:",
    average_precision_score(
        y_test,
        rf_probabilities
    )
)

print("\nConfusion Matrix:")

print(
    confusion_matrix(
        y_test,
        rf_predictions
    )
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        rf_predictions
    )
)


# =========================================================
# 15. SAVE MODELS
# =========================================================

joblib.dump(
    logistic_pipeline,
    "/home/sushil-singh/50_Projects/Pulse_prediction_V1/logistic_pipeline.pkl"
)

joblib.dump(
    random_forest_pipeline,
    "/home/sushil-singh/50_Projects/Pulse_prediction_V1/random_forest_pipeline.pkl"
)

print("\nModels saved successfully.")
```
